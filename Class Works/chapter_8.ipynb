```python
# 1. Who’s Weird? (Z-Score Ranking)
# • Compute z-scores for values
# • Rank items by how unusual they are
# • Label: Normal / Unusual / Extreme
# 2. Before vs After Normalization
# • Normalize data using z-score and min-max
# • Compare rankings before and after scaling
# 3. Feature Variability Ranking
# • Rank columns by standard deviation
# • Identify low-variance features
# 4. Design Your Own Dataset
# • Create a real-life numeric dataset
# • Normalize and summarize the data
# • Explain results in plain language
# Task 1: Who's Weird? (Z-Score Ranking)
print("=== Task 1: Who's Weird? (Z-Score Ranking) ===")
print("See cells below for z-score computation on Population")

# Task 2: Before vs After Normalization
print("\n=== Task 2: Before vs After Normalization ===")
# Original data
print("Before normalization:")
print(df[['MedInc', 'HouseAge']].head())

# Z-score normalization
df_zscore = (df - df.mean()) / df.std()
print("\nAfter Z-score normalization:")
print(df_zscore[['MedInc', 'HouseAge']].head())

# Min-Max normalization
df_minmax = (df - df.min()) / (df.max() - df.min())
print("\nAfter Min-Max normalization:")
print(df_minmax[['MedInc', 'HouseAge']].head())

# Task 3: Feature Variability Ranking
print("\n=== Task 3: Feature Variability Ranking ===")
variance_ranking = df.std().sort_values(ascending=False)
print("Features ranked by standard deviation:")
print(variance_ranking)

# Task 4: Design Your Own Dataset
print("\n=== Task 4: Design Your Own Dataset ===")
custom_data = pd.DataFrame({
    'StudentID': range(1, 11),
    'MathScore': [85, 92, 78, 95, 88, 76, 90, 82, 87, 91],
    'ReadingScore': [80, 88, 75, 90, 85, 72, 86, 79, 84, 89],
    'StudyHours': [5, 7, 4, 8, 6, 3, 7, 5, 6, 7]
})
print("\nCustom Dataset - Student Performance:")
print(custom_data)
print("\nSummary Statistics:")
print(custom_data.describe())

# Normalize custom dataset
custom_normalized = (custom_data[['MathScore', 'ReadingScore', 'StudyHours']] - 
                     custom_data[['MathScore', 'ReadingScore', 'StudyHours']].mean()) / \
                     custom_data[['MathScore', 'ReadingScore', 'StudyHours']].std()
print("\nNormalized Custom Dataset:")
print(custom_normalized)
```

$$\mu = \frac{\sum_{i=1}^{n} x_i}{n}$$

```python
# Using Pandas (Built on NumPy)
mean_income = df['MedInc'].mean()
house_age = df['HouseAge'].median()

print(f"Average Score: {mean_income}")
print(f"Middle Score: {house_age}")
```

$$\sigma = \sqrt{\frac{\sum(x_i - \mu)^2}{n}}$$

```python
# Using Pandas (Built on NumPy)
std_income = df['MedInc'].std()

print(f"Standard Deviation of income: {std_income}")
```

$z = \frac{x - \mu}{\sigma}$

```python
dfPop=df[['Population']].copy()
# print(dfPop['population'])
meanPop=dfPop.mean()
sdPop=dfPop.std()
# print(meanPop)
# print(sdPop)
dfPop['Z-score']=(dfPop-meanPop)/sdPop
# print(dfPop['Z-score'])
# print(dfPop)

# print(dfPop.head())

# dfPop['Z-score'].sort_values(ascending=False)
#             or
dfPop.sort_values(by='Z-score',ascending=False)
print(dfPop.head(10))
print(dfPop.describe())
```

```python
import numpy as np
#

dfPop['label'] = np.select(
    [
        dfPop['Z-score'].abs() < 2,
        dfPop['Z-score'].abs() < 3
    ],
    [
        'Normal',
        'Unusual'
    ],
    default='Extreme'
)
print(dfPop.head())
print(dfPop.describe())
print(dfPop['label'].value_counts())
```
